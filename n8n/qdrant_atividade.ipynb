```python
# generate query embedding
query_text = "gatos de pelo longo"

# search for similar menu items
results = client.query_points(
    collection_name="cat_items",
    query=Document(text=query_text, model="sentence-transformers/all-MiniLM-L6-v2"),
    with_payload=True,
    limit=5
)

#resultados printados
for result in results.points:
    print(f"Raça: {result.payload.get('raca', 'N/A')}")
    print(f"País: {result.payload.get('pais', 'N/A')}")
    print(f"Score: {result.score:.4f}")
    print(f"Pelagem: {result.payload.get('pelagem', 'N/A')}")
    print(f"Padrão de Cor: {result.payload.get('padrao', 'N/A')}")
    print("---")
```
